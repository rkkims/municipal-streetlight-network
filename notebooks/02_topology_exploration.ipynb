```python
import geopandas as gpd
import matplotlib.pyplot as plt

network_id = "LGTNET00001"

poles = gpd.read_file("data/normalized/dnv/poles.gpkg")
conduit = gpd.read_file("data/normalized/dnv/conduit.gpkg")
fittings = gpd.read_file("data/normalized/dnv/fittings.gpkg")

poles_selected = poles[poles["network_id"] == network_id]
conduit_selected = conduit[conduit["network_id"] == network_id]
fittings_selected = fittings[fittings["network_id"] == network_id]

ax = conduit_selected.plot(color="black", linewidth=2, figsize=(10, 10))
poles_selected.plot(ax=ax, color="red", markersize=20, label="Poles")
fittings_selected.plot(ax=ax, color="blue", markersize=25, label="Fittings")

plt.legend()
plt.show()
```
